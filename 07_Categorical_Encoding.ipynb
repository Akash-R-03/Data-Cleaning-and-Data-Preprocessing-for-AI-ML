# 1. What is Categorical Encoding?

## Concept

Categorical encoding is the process of converting categorical values into numerical values so that Machine Learning algorithms can process them.

In the Titanic dataset, columns such as `Sex`, `Embarked`, and `Cabin` contain categorical information. Most Machine Learning algorithms work with numerical data, so these values may need to be encoded.

### Example

For example:

- Male → 0
- Female → 1

This converts a categorical variable into a numerical representation.

### Implementation

We will first load the Titanic dataset and identify the categorical columns.

### Result

The categorical columns in the Titanic dataset can be identified before selecting an appropriate encoding technique.

### Interpretation

Different categorical columns may require different encoding methods depending on whether they are nominal, ordinal, binary, or high-cardinality.

### Insight

There is no single encoding technique that is suitable for every categorical variable.

### AI/ML Relevance

Encoding converts categorical information into a numerical form that Machine Learning algorithms can process.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

categorical_cols = df.select_dtypes(include=["object", "string"]).columns

print("Categorical Columns:")
print(list(categorical_cols))
```

# 2. What are Nominal Variables?

## Concept

Nominal variables are categorical variables whose values do not have a natural order or ranking.

### Example

In the Titanic dataset:

- `Sex` → male, female
- `Embarked` → S, C, Q

There is no meaningful order between these categories.

### Implementation

We can examine the unique values in the `Sex` and `Embarked` columns.

### Result

The output shows the different categories available in these columns.

### Interpretation

Since these categories do not have a meaningful ranking, ordinal encoding should generally not be used for them.

### Insight

Nominal variables are usually suitable for techniques such as One-Hot Encoding.

### AI/ML Relevance

Correctly encoding nominal variables prevents the model from incorrectly assuming that one category is greater or smaller than another.

```python
print("Sex categories:")
print(df["Sex"].dropna().unique())

print("\nEmbarked categories:")
print(df["Embarked"].dropna().unique())
```

# 3. What are Ordinal Variables?

## Concept

Ordinal variables are categorical variables whose values have a meaningful order or ranking.

The Titanic dataset does not contain a clear ordinal feature such as education level or satisfaction level.

### Example

An ordinal variable could be:

- Low
- Medium
- High

Here, High has a greater order than Medium, and Medium has a greater order than Low.

### Implementation

We can create a small example to understand ordinal encoding.

### Result

The categories are converted according to their meaningful order.

### Interpretation

Ordinal encoding is appropriate only when the categories have a natural ranking.

### Insight

We should not create an artificial order for nominal variables.

### AI/ML Relevance

Using ordinal encoding for a nominal variable can introduce false relationships between categories and negatively affect model performance.

```python
ordinal_data = pd.DataFrame({
    "Level": ["Low", "Medium", "High", "Medium", "Low"]
})

order = {"Low": 1, "Medium": 2, "High": 3}

ordinal_data["Level_Encoded"] = ordinal_data["Level"].map(order)

print(ordinal_data)
```

# 4. What are Binary Variables?

## Concept

A binary variable contains only two possible categories.

### Example

The `Sex` column in the Titanic dataset contains two main categories:

- male
- female

This makes it suitable for binary encoding.

### Implementation

We can convert the `Sex` column into numerical values.

### Result

Male and female are represented using two numerical values.

### Interpretation

Binary variables can usually be represented using 0 and 1 without creating multiple columns.

### Insight

Simple binary encoding is useful when a categorical feature contains exactly two categories.

### AI/ML Relevance

Converting binary categories into numerical values allows Machine Learning algorithms to use the feature directly.

```python
df_binary = df[["Sex"]].dropna().copy()

df_binary["Sex_Encoded"] = df_binary["Sex"].map({
    "male": 0,
    "female": 1
})

print(df_binary.head())
```

# 5. What is Label Encoding?

## Concept

Label Encoding assigns an integer value to each category.

For example:

- Category A → 0
- Category B → 1
- Category C → 2

### Example

The `Embarked` column contains categories such as S, C, and Q.

### Implementation

We can use `LabelEncoder` to convert these categories into numerical values.

### Result

Each category receives an integer value.

### Interpretation

Label encoding is simple and memory efficient, but the numerical values may appear to have an order even when no real order exists.

### Insight

Label encoding should be used carefully for nominal variables. It is more appropriate when the encoded values have a meaningful interpretation or for certain tree-based models.

### AI/ML Relevance

Incorrect use of label encoding can introduce an artificial relationship between categories.

```python
from sklearn.preprocessing import LabelEncoder

data = df[["Embarked"]].dropna().copy()

encoder = LabelEncoder()
data["Embarked_Encoded"] = encoder.fit_transform(data["Embarked"])

print(data.head())
```

# 6. What is Ordinal Encoding?

## Concept

Ordinal Encoding converts ordered categories into numerical values according to their natural ranking.

### Example

For example:

- Low → 1
- Medium → 2
- High → 3

### Implementation

The Titanic dataset does not contain a clear ordinal feature, so we will use a simple example to demonstrate the technique.

### Result

The ordered categories are converted into numerical values while preserving their ranking.

### Interpretation

The numerical values represent the original order of the categories.

### Insight

Ordinal Encoding should only be used when a genuine order exists.

### AI/ML Relevance

Preserving meaningful order can help a Machine Learning model learn relationships between ordinal categories.

```python
from sklearn.preprocessing import OrdinalEncoder

example = pd.DataFrame({
    "Class_Level": ["Low", "Medium", "High", "Medium"]
})

encoder = OrdinalEncoder(categories=[["Low", "Medium", "High"]])

example["Encoded"] = encoder.fit_transform(example[["Class_Level"]])

print(example)
```

# 7. What is One-Hot Encoding?

## Concept

One-Hot Encoding creates a separate binary column for each category.

### Example

For the `Embarked` column:

- C
- Q
- S

One-Hot Encoding creates separate columns for these categories.

### Implementation

We can use pandas `get_dummies()` to encode the `Embarked` column.

### Result

Each category becomes a separate column containing 0 or 1.

### Interpretation

A value of 1 indicates that the record belongs to that category.

### Insight

One-Hot Encoding is useful for nominal variables because it does not create an artificial ranking between categories.

### AI/ML Relevance

It allows Machine Learning models to use categorical information without incorrectly assuming that one category is greater than another.

```python
encoded_embarked = pd.get_dummies(
    df["Embarked"],
    prefix="Embarked",
    dtype=int
)

print(encoded_embarked.head())
```

# 8. What are Dummy Variables?

## Concept

Dummy variables are binary variables created from categorical variables.

For a categorical variable with multiple categories, dummy encoding represents category membership using 0 and 1.

### Example

For the `Embarked` column, dummy variables can represent whether a passenger embarked from C, Q, or S.

### Implementation

We can create dummy variables using `pd.get_dummies()`.

### Result

The categorical column is represented using binary columns.

### Interpretation

Each row indicates whether the passenger belongs to a particular category.

### Insight

When using a linear model, one dummy column can be dropped to avoid perfect multicollinearity.

### AI/ML Relevance

Dummy variables allow categorical information to be included in numerical Machine Learning models.

```python
dummy_data = pd.get_dummies(
    df["Embarked"],
    prefix="Embarked",
    drop_first=True,
    dtype=int
)

print(dummy_data.head())
```

# 9. What is Frequency Encoding?

## Concept

Frequency Encoding replaces each category with the number of times it occurs in the dataset, or its frequency.

### Example

If one category occurs 500 times and another occurs 200 times, their encoded values can represent these frequencies.

### Implementation

We will apply frequency encoding to the `Embarked` column.

### Result

Each category is replaced by its frequency in the dataset.

### Interpretation

The encoded value represents how common that category is.

### Insight

Frequency Encoding can be useful when a categorical feature contains many unique categories.

### AI/ML Relevance

It can reduce the number of columns compared with One-Hot Encoding, especially for high-cardinality variables.

```python
frequency = df["Embarked"].value_counts()

df_frequency = df[["Embarked"]].copy()
df_frequency["Embarked_Frequency"] = df_frequency["Embarked"].map(frequency)

print(df_frequency.head())
```

# 10. What is Target Encoding?

## Concept

Target Encoding replaces each category with a statistic calculated from the target variable, commonly the mean target value for that category.

### Example

In the Titanic dataset, `Survived` can be used as the target.

We can calculate the average survival rate for each `Embarked` category.

### Implementation

We calculate the mean `Survived` value for each `Embarked` category and map it back to the data.

### Result

Each category receives a value representing its average target outcome.

### Interpretation

For example, a value of 0.60 means that approximately 60% of passengers in that category survived in the data used for calculating the encoding.

### Insight

Target Encoding must be performed carefully because it uses target information.

### AI/ML Relevance

If target encoding is calculated using the complete dataset before train-test splitting, it can cause data leakage. In real projects, the encoding should be learned from the training data only.

```python
target_mean = df.groupby("Embarked")["Survived"].mean()

df_target = df[["Embarked"]].copy()
df_target["Embarked_Target_Encoded"] = df_target["Embarked"].map(target_mean)

print(df_target.head())
```

# 11. How do we Handle Unknown Categories?

## Concept

An unknown category occurs when a new category appears during prediction that was not present when the encoder was trained.

### Example

Suppose the training data contains:

- C
- Q
- S

But new data contains another category that was not seen during training.

### Implementation

`OneHotEncoder` can be configured to ignore unknown categories.

### Result

The model can process new categories without producing an encoding error.

### Interpretation

Unknown categories will not create a new unexpected column during transformation.

### Insight

Handling unknown categories is important when a model is used on future or production data.

### AI/ML Relevance

Proper handling of unknown categories makes preprocessing more robust and prevents prediction-time errors.

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

encoded = encoder.fit_transform(df[["Embarked"]].fillna("Missing"))

print(encoded[:5])
```

# 12. What is High-Cardinality Categorical Data?

## Concept

High-cardinality categorical data contains a large number of unique categories.

### Example

The `Ticket` and `Cabin` columns in the Titanic dataset contain many different values compared with columns such as `Sex`.

### Implementation

We can count the number of unique values in these columns.

### Result

The output shows that `Ticket` and `Cabin` have many unique categories.

### Interpretation

Applying One-Hot Encoding directly to such columns can create a large number of features.

### Insight

High-cardinality features may require alternative approaches such as frequency encoding, target encoding, grouping rare categories, or extracting useful information from the original values.

### AI/ML Relevance

Poor handling of high-cardinality variables can increase memory usage, model complexity, and the risk of overfitting.

```python
high_cardinality_cols = ["Ticket", "Cabin"]

for col in high_cardinality_cols:
    print(col, "unique values:", df[col].nunique())
```

# 13. When Should Each Encoding Technique Be Used?

## Concept

The encoding technique should be selected based on the type of categorical variable, number of categories, and Machine Learning model.

### Encoding Guide

| Encoding Technique | Suitable Use |
|---|---|
| Label Encoding | Simple categorical-to-integer conversion; use carefully for nominal data |
| Ordinal Encoding | Categories with a meaningful order |
| One-Hot Encoding | Nominal variables with a small or moderate number of categories |
| Dummy Variables | One-Hot Encoding with one category removed, commonly useful for linear models |
| Frequency Encoding | Categories where frequency provides useful information, especially higher cardinality |
| Target Encoding | Categories with useful target-based relationships; must avoid data leakage |
| Binary Encoding | Useful when categorical variables have many categories |
| Handling Unknown Categories | Important when new categories may appear in future data |

### Example

For the Titanic dataset:

- `Sex` → Binary Encoding can be suitable.
- `Embarked` → One-Hot Encoding is suitable.
- `Ticket` → High-cardinality handling may be more appropriate than direct One-Hot Encoding.
- A genuinely ordered feature → Ordinal Encoding.

### Insight

The best encoding method depends on the data and the Machine Learning problem rather than simply choosing one method for all categorical columns.

### AI/ML Relevance

Choosing the correct encoding helps preserve useful information and reduces unnecessary complexity in the Machine Learning pipeline.

```python
print("Sex unique values:", df["Sex"].nunique())
print("Embarked unique values:", df["Embarked"].nunique())
print("Ticket unique values:", df["Ticket"].nunique())
print("Cabin unique values:", df["Cabin"].nunique())
```

# 14. What are the Risks of Inappropriate Encoding?

## Concept

Using an inappropriate encoding technique can introduce incorrect relationships, increase the number of features, cause data leakage, or make the model unnecessarily complex.

### Problem

A categorical variable may be encoded in a way that does not represent its actual meaning.

### Analysis

For example, assigning:

- C → 0
- Q → 1
- S → 2

may make a model interpret S as greater than Q and Q as greater than C, even though these categories have no natural order.

### Technique Selected

For nominal variables such as `Embarked`, One-Hot Encoding is generally more appropriate.

### Reason

It represents each category independently without creating an artificial ranking.

### Implementation

We can compare the original categorical column with its One-Hot encoded representation.

### Result

The categories are converted into separate binary features.

### Impact

Appropriate encoding can improve the quality of model inputs, while inappropriate encoding can introduce bias, misleading relationships, data leakage, or unnecessary model complexity.

### Insight

Encoding is not just a technical conversion. It is a preprocessing decision that should be based on the meaning and structure of the data.

### AI/ML Relevance

Correct categorical encoding helps Machine Learning algorithms learn meaningful patterns from categorical features.

```python
encoded_data = pd.get_dummies(
    df["Embarked"],
    prefix="Embarked",
    dtype=int
)

print("Original values:")
print(df["Embarked"].head())

print("\nEncoded values:")
print(encoded_data.head())
```
