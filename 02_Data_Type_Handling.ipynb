# 7. Converting Data Types

## Concept

Sometimes a column is stored using an incorrect data type. Data type conversion changes it into the appropriate format.

## Example

A numeric column stored as text should be converted into a numerical type before performing calculations.

## Implementation

We can use Pandas functions such as `astype()`, `to_numeric()`, and `to_datetime()`.

## Output

The data type before and after conversion can be compared.

## Interpretation

The conversion confirms that the column is now stored using the required data type.

## Insight

Correct conversion makes the data easier to analyze and prevents calculation errors.

## AI/ML Relevance

Machine Learning preprocessing depends on correctly formatted features, so data type conversion is an important preprocessing step.

```python
print("Original Age Data Type:")
print(df["Age"].dtype)

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

print("\nUpdated Age Data Type:")
print(df["Age"].dtype)
```

# 8. astype()

## Concept

## Example

## Implementation

## Output

# 9. to_numeric()

## Concept

`pd.to_numeric()` converts values into numerical form.

## Example

If a column contains numbers stored as strings, it can be converted into numeric values.

## Implementation

We use `errors="coerce"` so that invalid values are converted to missing values instead of causing an error.

## Output

The converted column and its data type are displayed.

## Interpretation

Valid numerical values remain numeric, while invalid values become `NaN`.

## Insight

This method is useful when datasets contain mixed or inconsistent numerical values.

## AI/ML Relevance

Numerical features must be properly converted before statistical analysis and Machine Learning.

```python
age_numeric = pd.to_numeric(df["Age"], errors="coerce")

print(age_numeric.head())
print("\nData Type:", age_numeric.dtype)
```

# 10. to_datetime()

## Concept

`pd.to_datetime()` converts values into a proper date/time format.

## Example

For example, a value such as `"2026-08-25"` stored as text can be converted into a datetime value.

The Titanic dataset does not contain a date column, so we demonstrate the method using a small example.

## Implementation

We convert string dates into datetime values.

## Output

The converted values and their data type are displayed.

## Interpretation

The values are now recognized as dates instead of normal strings.

## Insight

Proper datetime conversion makes it possible to extract useful information such as year, month, and day.

## AI/ML Relevance

Date-based features can be transformed into meaningful numerical features for Machine Learning.

```python
dates = pd.Series(["2026-08-25", "2026-08-26"])

dates = pd.to_datetime(dates)

print(dates)
print("\nData Type:", dates.dtype)
```

# 11. Handling Invalid Conversions

## Concept

Real-world datasets may contain invalid values that cannot be converted into the required data type.

## Example

A numerical column may contain values such as `"unknown"` or `"not available"`.

## Implementation

Using `errors="coerce"` converts invalid values into `NaN`.

## Output

Invalid values are shown as missing values.

## Interpretation

This prevents the program from stopping because of conversion errors.

## Insight

Invalid values should not simply be ignored. They should be identified and handled during later preprocessing.

## AI/ML Relevance

Handling invalid values prevents data errors and helps maintain reliable input for Machine Learning models.

```python
example_values = pd.Series(["100", "250", "unknown", "500"])

converted = pd.to_numeric(example_values, errors="coerce")

print(converted)
```
