# 1. What is Feature Selection?

### Question: What is Feature Selection?

Feature Selection is the process of selecting the most useful features from a dataset for Machine Learning.

The goal is to keep features that provide useful information and remove features that are irrelevant or unnecessary.

### Example

In the Titanic dataset, features such as `Pclass`, `Sex`, `Age`, and `Fare` may provide useful information for predicting passenger survival.

### Insight

Feature selection helps us focus on the features that are most useful for the Machine Learning problem.

### AI/ML Relevance

Selecting useful features can reduce unnecessary information and make Machine Learning models simpler and more efficient.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)
print("\nFeatures:")
print(df.columns.tolist())
```

# 2. Why Feature Selection?

### Question: Why is Feature Selection important?

Feature selection is important because a dataset may contain irrelevant, redundant, or unnecessary features.

Removing such features can:

- Reduce model complexity.
- Reduce training time.
- Reduce noise.
- Improve model interpretability.
- Help reduce the risk of overfitting.

### Insight

More features do not always mean a better Machine Learning model.

### AI/ML Relevance

A model should use features that provide useful information for predicting the target variable.

```python
print("Number of features before selection:", df.shape[1])

print("\nFeature names:")
print(df.columns.tolist())
```

# 3. Relevant Features

### Question: What are Relevant Features?

Relevant features are features that contain useful information related to the target variable.

### Example

For Titanic survival prediction, `Pclass`, `Sex`, `Age`, and `Fare` can be considered potentially relevant because they may contain information related to passenger survival.

### Insight

A relevant feature can contribute useful information to the prediction.

### AI/ML Relevance

Keeping relevant features can help the model learn meaningful patterns from the data.

```python
target = "Survived"

relevant_features = ["Pclass", "Sex", "Age", "Fare"]

print("Potentially relevant features:")
print(relevant_features)

print("\nTarget:", target)
```

# 4. Irrelevant Features

### Question: What are Irrelevant Features?

Irrelevant features are features that provide little or no useful information for the target prediction.

### Example

`PassengerId` is mainly an identifier. It does not represent a meaningful passenger characteristic and may not be useful for predicting survival.

### Insight

Identifier columns should be examined carefully before using them as Machine Learning features.

### AI/ML Relevance

Removing irrelevant features can reduce noise and prevent the model from learning meaningless patterns.

```python
print("PassengerId unique values:", df["PassengerId"].nunique())
print("Total rows:", len(df))

print("\nPassengerId can be considered an identifier rather than a useful predictive feature.")
```

# 5. Redundant Features

### Question: What are Redundant Features?

Redundant features contain information that is very similar to information already provided by another feature.

Highly correlated numerical features can sometimes be redundant.

### Example

If two features contain almost the same information, keeping both may not provide much additional value.

### Insight

Redundant features can increase model complexity without adding much new information.

### AI/ML Relevance

Removing unnecessary redundancy can make a model simpler and easier to interpret.

```python
numeric_df = df.select_dtypes(include="number")

correlation_matrix = numeric_df.corr()

print(correlation_matrix)
```

# 6. Correlation-Based Feature Selection

### Question: What is Correlation-Based Feature Selection?

Correlation-based selection uses the correlation between numerical features to identify highly related or redundant features.

A correlation value close to:

- +1 indicates a strong positive relationship.
- -1 indicates a strong negative relationship.
- 0 indicates little linear relationship.

Correlation can help identify features that provide similar information.

### Insight

Correlation is useful for checking relationships, but it should not be the only method used for feature selection.

### AI/ML Relevance

Correlation analysis can help identify redundant numerical features before model training.

```python
correlation = df.select_dtypes(include="number").corr()

print("Correlation with Survived:")
print(correlation["Survived"].sort_values(ascending=False))
```

# 7. Variance Threshold

### Question: What is Variance Threshold?

Variance Threshold is a feature selection technique that removes features with very low variance.

A feature with almost the same value for every observation provides very little variation.

### Example

If a column contains nearly the same value for all passengers, it may not provide useful information for the model.

### Insight

Low-variance features can sometimes be removed because they contain limited information.

### AI/ML Relevance

Variance Threshold can reduce the number of features before model training.

```python
from sklearn.feature_selection import VarianceThreshold

numeric_features = df.select_dtypes(include="number").drop(columns=["Survived"])

selector = VarianceThreshold(threshold=0.01)

selector.fit(numeric_features)

selected_features = numeric_features.columns[selector.get_support()]

print("Selected features:")
print(selected_features.tolist())
```

# 8. Univariate Feature Selection

### Question: What is Univariate Feature Selection?

Univariate feature selection evaluates each feature individually based on its relationship with the target.

For classification problems, statistical tests such as ANOVA can be used for suitable numerical features.

### Example

For Titanic survival prediction, numerical features can be evaluated individually against `Survived`.

### Insight

A feature can be selected based on how strongly it is associated with the target when considered individually.

### AI/ML Relevance

Univariate selection provides a simple way to identify potentially useful features before model training.

```python
from sklearn.feature_selection import SelectKBest, f_classif

X = df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].fillna(
    df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].median()
)

y = df["Survived"]

selector = SelectKBest(score_func=f_classif, k=3)

X_selected = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]

print("Top selected features:")
print(selected_features.tolist())
```

# 9. Mutual Information

### Question: What is Mutual Information?

Mutual Information measures how much information one feature provides about the target.

A higher mutual information score generally indicates a stronger dependency between the feature and the target.

### Insight

Mutual Information can detect relationships that are not necessarily simple linear relationships.

### AI/ML Relevance

It can help identify informative features for Machine Learning models.

```python
from sklearn.feature_selection import mutual_info_classif

X = df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].copy()
X = X.fillna(X.median())

y = df["Survived"]

scores = mutual_info_classif(X, y, random_state=42)

mi_scores = pd.Series(scores, index=X.columns).sort_values(
    ascending=False
)

print("Mutual Information Scores:")
print(mi_scores)
```

# 10. Recursive Feature Elimination (RFE)

### Question: What is Recursive Feature Elimination?

Recursive Feature Elimination (RFE) is a feature selection technique that repeatedly removes less important features based on a Machine Learning model.

The process continues until the required number of features remains.

### Insight

RFE selects features based on their contribution to a model rather than looking at each feature independently.

### AI/ML Relevance

RFE can help create a smaller set of features that works well with a selected Machine Learning algorithm.

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

X = df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].fillna(
    df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].median()
)

y = df["Survived"]

model = LogisticRegression(max_iter=1000)

rfe = RFE(model, n_features_to_select=3)

rfe.fit(X, y)

print("Selected features:")
print(X.columns[rfe.support_].tolist())
```

# 11. Feature Importance

### Question: What is Feature Importance?

Feature importance shows how useful each feature is for making predictions in a particular Machine Learning model.

Tree-based models such as Random Forest can provide feature importance scores.

### Example

We can use Random Forest to estimate the importance of selected Titanic features for predicting `Survived`.

### Insight

Feature importance helps us understand which features contribute more to the model's predictions.

### AI/ML Relevance

Feature importance improves model interpretability and can support feature selection decisions.

```python
from sklearn.ensemble import RandomForestClassifier

X = df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].fillna(
    df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].median()
)

y = df["Survived"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

print("Feature Importance:")
print(importance)
```

# 12. How Can Unnecessary Features Affect Machine Learning Models?

### Question: How do unnecessary features affect Machine Learning models?

Unnecessary features can:

- Add noise to the dataset.
- Increase model complexity.
- Increase training time.
- Make the model harder to interpret.
- Increase the risk of overfitting.
- Reduce model efficiency.

However, a feature should not be removed simply because it appears less important from one analysis. The decision should be supported by the data and the Machine Learning objective.

### Insight

Feature selection should be based on evidence rather than blindly removing columns.

### AI/ML Relevance

Using useful features can help create simpler, more efficient, and more interpretable Machine Learning models.
