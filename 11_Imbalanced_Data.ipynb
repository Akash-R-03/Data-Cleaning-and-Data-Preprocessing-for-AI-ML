# 1. What is Class Imbalance?

### Question: What is Class Imbalance?

Class imbalance occurs when the classes in a classification dataset are not represented equally.

For example, in the Titanic dataset, `Survived` has two classes:

- 0 → Did not survive
- 1 → Survived

If one class has many more records than the other, the dataset is considered imbalanced.

### Example

If 800 passengers did not survive and only 300 passengers survived, the two classes are not balanced.

### Insight

Before training a classification model, we should check how many observations belong to each class.

### AI/ML Relevance

Class imbalance can affect how a Machine Learning model learns and can cause it to favor the majority class.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

print("Target Class Counts:")
print(df["Survived"].value_counts())
```

# 2. Balanced vs Imbalanced Dataset

### Question: What is the difference between a balanced and imbalanced dataset?

A balanced dataset has approximately similar numbers of observations in each class.

An imbalanced dataset has a noticeable difference between the number of observations in different classes.

### Example

A dataset with:

- Class 0 → 500 records
- Class 1 → 500 records

is balanced.

A dataset with:

- Class 0 → 800 records
- Class 1 → 200 records

is imbalanced.

### Insight

The Titanic target variable can be checked to understand whether the classes are balanced.

### AI/ML Relevance

Knowing the class distribution helps us decide whether an imbalance-handling technique is required.

```python
class_counts = df["Survived"].value_counts()

print(class_counts)

print("\nClass Percentages:")
print(df["Survived"].value_counts(normalize=True) * 100)
```

# 3. Why is Class Imbalance a Problem?

### Question: Why can class imbalance be a problem?

When one class is much larger than another, a Machine Learning model may become biased toward the majority class.

The model may perform well on the majority class but perform poorly on the minority class.

### Example

If 90% of passengers belong to one class, a model that predicts the majority class most of the time can still achieve high accuracy.

### Insight

We should not judge an imbalanced classification model using accuracy alone.

### AI/ML Relevance

Class imbalance can reduce the model's ability to correctly identify minority-class observations.

```python
majority_percentage = df["Survived"].value_counts(normalize=True).max() * 100

print(f"Majority class percentage: {majority_percentage:.2f}%")
```

# 4. Undersampling

### Question: What is Undersampling?

Undersampling reduces the number of observations in the majority class so that it becomes closer to the minority class.

### Example

If Class 0 has 600 records and Class 1 has 300 records, some Class 0 records can be removed to create a more balanced dataset.

### When to use

Undersampling can be useful when the dataset is large and there are enough majority-class observations.

### Advantage

It reduces the size of the dataset and can make the classes more balanced.

### Limitation

Some useful information from the majority class may be lost.

### AI/ML Relevance

Undersampling can help prevent a model from becoming too focused on the majority class.

```python
from sklearn.utils import resample

majority = df[df["Survived"] == 0]
minority = df[df["Survived"] == 1]

majority_downsampled = resample(
    majority,
    replace=False,
    n_samples=len(minority),
    random_state=42
)

undersampled_df = pd.concat([majority_downsampled, minority])

print(undersampled_df["Survived"].value_counts())
```

# 5. Oversampling

### Question: What is Oversampling?

Oversampling increases the number of minority-class observations so that the classes become more balanced.

### Example

If Class 0 has 600 records and Class 1 has 300 records, minority-class observations can be increased to match the majority class.

### When to use

Oversampling is useful when the minority class contains important information and we do not want to remove majority-class observations.

### Advantage

It keeps the available majority-class observations.

### Limitation

Simple oversampling can duplicate observations and may increase the risk of overfitting.

### AI/ML Relevance

Oversampling can help a model learn more effectively from the minority class.

```python
minority_upsampled = resample(
    minority,
    replace=True,
    n_samples=len(majority),
    random_state=42
)

oversampled_df = pd.concat([majority, minority_upsampled])

print(oversampled_df["Survived"].value_counts())
```

# 6. Random Oversampling

### Question: What is Random Oversampling?

Random Oversampling randomly duplicates observations from the minority class until the desired class balance is achieved.

### Example

Minority-class passenger records can be randomly selected more than once to increase their representation.

### When to use

Use it when a simple and easy-to-understand balancing technique is required.

### Advantage

It is simple and does not remove majority-class observations.

### Limitation

Duplicating the same observations can increase the risk of overfitting.

### AI/ML Relevance

It can give the model more exposure to minority-class examples during training.

```python
!pip install imbalanced-learn
```

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset loaded successfully.")
print(df.shape)
```

```python
from imblearn.over_sampling import RandomOverSampler

X = df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].copy()
X = X.fillna(X.median())

y = df["Survived"]

ros = RandomOverSampler(random_state=42)

X_over, y_over = ros.fit_resample(X, y)

print("Original:")
print(y.value_counts())

print("\nAfter Random Oversampling:")
print(y_over.value_counts())
```

# 7. Random Undersampling

### Question: What is Random Undersampling?

Random Undersampling randomly removes observations from the majority class until the classes are balanced.

### When to use

It is useful when the dataset contains a large number of majority-class observations.

### Advantage

It reduces the dataset size and balances the classes quickly.

### Limitation

Useful information may be removed from the majority class.

### AI/ML Relevance

It can reduce majority-class bias, but the loss of information should be considered before using it.

```python
from imblearn.under_sampling import RandomUnderSampler

rus = RandomUnderSampler(random_state=42)

X_under, y_under = rus.fit_resample(X, y)

print("Original:")
print(y.value_counts())

print("\nAfter Random Undersampling:")
print(y_under.value_counts())
```

# 8. SMOTE

### Question: What is SMOTE?

SMOTE stands for Synthetic Minority Over-sampling Technique.

Instead of simply duplicating minority-class records, SMOTE creates new synthetic observations based on existing minority-class observations.

### When to use

SMOTE can be useful when the minority class is too small and simple duplication may cause overfitting.

### Advantage

It creates new synthetic minority observations instead of simply copying existing records.

### Limitation

Synthetic observations may not always represent realistic data.

### AI/ML Relevance

SMOTE can help a classification model learn a better representation of the minority class.

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_smote, y_smote = smote.fit_resample(X, y)

print("Original:")
print(y.value_counts())

print("\nAfter SMOTE:")
print(y_smote.value_counts())
```

# 9. Borderline-SMOTE

### Question: What is Borderline-SMOTE?

Borderline-SMOTE is a variation of SMOTE that focuses mainly on minority-class observations located near the decision boundary.

These observations can be more difficult for a classifier to distinguish.

### When to use

It can be useful when minority-class observations near the class boundary are especially important.

### Advantage

It focuses synthetic sampling on difficult minority observations.

### Limitation

It can introduce synthetic observations in areas where the classes overlap.

### AI/ML Relevance

Borderline-SMOTE can help a model learn difficult boundaries between classes.

```python
from imblearn.over_sampling import BorderlineSMOTE

borderline_smote = BorderlineSMOTE(random_state=42)

X_border, y_border = borderline_smote.fit_resample(X, y)

print("Original:")
print(y.value_counts())

print("\nAfter Borderline-SMOTE:")
print(y_border.value_counts())
```

# 10. Class Weights

### Question: What are Class Weights?

Class weights assign different importance to different classes during model training.

The minority class can be given a higher weight so that mistakes on that class have a greater effect on the model.

### Example

A classification model can use `class_weight="balanced"` to automatically give more importance to the minority class.

### When to use

Class weights are useful when we want to handle imbalance without changing the original dataset.

### Advantage

No observations need to be duplicated or removed.

### Limitation

Not every Machine Learning algorithm supports class weights.

### AI/ML Relevance

Class weights can help a model pay more attention to minority-class observations.

```python
from sklearn.linear_model import LogisticRegression

X = df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].fillna(
    df[["Pclass", "Age", "SibSp", "Parch", "Fare"]].median()
)

y = df["Survived"]

model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000
)

model.fit(X, y)

print("Logistic Regression trained using balanced class weights.")
```

# 11. When Should Each Technique Be Used?

### Question: When should we use different imbalance-handling techniques?

The technique should depend on the dataset size, class distribution, and Machine Learning model.

- **Random Oversampling:** Simple option when minority observations need to be increased.
- **Random Undersampling:** Useful when the majority class is very large.
- **SMOTE:** Useful when synthetic minority observations are appropriate.
- **Borderline-SMOTE:** Useful when difficult boundary observations need more attention.
- **Class Weights:** Useful when we want to keep the original dataset.
- **No technique:** If the class distribution is reasonably balanced, additional balancing may not be necessary.

### Insight

There is no single technique that is always best. The method should be selected based on the actual problem.

### AI/ML Relevance

Choosing an appropriate technique can improve minority-class performance without unnecessarily changing useful information.

```python
print("Original class distribution:")
print(y.value_counts())

print("\nOriginal class percentages:")
print((y.value_counts(normalize=True) * 100).round(2))
```

# 12. Why Accuracy Alone Can Be Misleading

### Question: Why can accuracy be misleading for an imbalanced dataset?

Accuracy measures the overall percentage of correct predictions.

In an imbalanced dataset, a model can achieve high accuracy simply by predicting the majority class most of the time.

### Example

Suppose 95% of observations belong to Class 0 and only 5% belong to Class 1.

A model that predicts Class 0 for every observation would have 95% accuracy, but it would completely fail to identify Class 1.

### Insight

For imbalanced classification problems, we should also consider metrics such as:

- Precision
- Recall
- F1-score
- Confusion Matrix
- ROC-AUC

### AI/ML Relevance

Using multiple evaluation metrics gives a better understanding of how well the model handles both majority and minority classes.

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, classification_report

dummy = DummyClassifier(strategy="most_frequent")

dummy.fit(X, y)

predictions = dummy.predict(X)

print("Accuracy:", accuracy_score(y, predictions))

print("\nClassification Report:")
print(classification_report(y, predictions, zero_division=0))
```
