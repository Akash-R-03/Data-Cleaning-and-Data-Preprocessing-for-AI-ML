# Missing Value Handling

## 1. What are Missing Values?

Missing values occur when information is not available for a particular observation or column.

In the Titanic dataset, some columns contain missing values. For example, `Age`, `Cabin`, and `Embarked` have missing records.

### Example

If a passenger's age is not recorded, the `Age` value will be missing.

### Interpretation

Missing values can affect statistical analysis and Machine Learning models if they are not handled properly.

### Insight

Before choosing an imputation technique, we should first understand how many values are missing and why they may be missing.

### AI/ML Relevance

Most Machine Learning algorithms cannot directly work with missing values, so missing-value handling is an important preprocessing step.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())
```

## 2. Why Missing Values Occur

Missing values can occur for different reasons.

### MCAR – Missing Completely At Random

The missing value is unrelated to any other variable.

Example: Some records may be lost because of a technical issue during data collection.

### MAR – Missing At Random

The missing value is related to another observed variable.

Example: Age information may be missing more often for passengers from a particular group.

### MNAR – Missing Not At Random

The missingness is related to the missing value itself.

Example: People with very high income may be less likely to report their income.

### Interpretation

Understanding the reason for missing values helps us select an appropriate treatment method.

### Insight

There is no single imputation technique that is suitable for every missing-value situation.

### AI/ML Relevance

Incorrectly handling missing values can introduce bias and negatively affect model performance.

```python
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": df.isnull().mean() * 100
})

print(missing_summary[missing_summary["Missing Count"] > 0])
```

## 3. Detecting Missing Values

### Concept

We can use `isnull()` or `isna()` to identify missing values.

### Example

If `Age` contains missing values, `df["Age"].isnull().sum()` gives the number of missing records.

### Implementation

We will check both the missing count and missing percentage for every column.

### Interpretation

Columns with a higher missing percentage require more attention during preprocessing.

### Insight

The missing-value percentage gives us a better understanding of the severity of the problem than the count alone.

### AI/ML Relevance

Missing-value analysis helps us decide whether to impute, remove, or retain a feature before model training.

```python
missing_count = df.isnull().sum()
missing_percentage = df.isnull().mean() * 100

missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print(missing_report[missing_report["Missing Count"] > 0])
```

## 4. Dropping Rows

### Concept

Rows containing missing values can be removed using `dropna()`.

### Example

If only a very small number of passenger records have missing values, removing those records may be acceptable.

### When to Use

- When only a small number of rows are affected.
- When the missing records are not important.
- When removing them will not introduce significant bias.

### When Not to Use

- When many rows contain missing values.
- When the dataset is already small.
- When removing rows may change the data distribution.

### Advantages

- Simple and easy to implement.
- Does not create artificial values.

### Limitations

- Reduces the number of observations.
- Important information may be lost.

### Interpretation

We should not remove rows automatically. The percentage and importance of missing records should be considered first.

### Insight

Dropping rows is useful only when the amount of missing data is relatively small.

### AI/ML Relevance

Removing too many rows can reduce training data and may negatively affect model performance.

```python
df_drop_rows = df.dropna()

print("Original rows:", len(df))
print("Rows after dropping missing values:", len(df_drop_rows))
```

## 5. Dropping Columns

### Concept

A column can be removed when it contains too many missing values or provides little useful information.

### Example

The `Cabin` column contains many missing values in the Titanic dataset.

### When to Use

- When a column has a very high percentage of missing values.
- When the feature is not useful for the analysis.
- When reliable imputation is not possible.

### When Not to Use

- When the feature is important.
- When useful information can be extracted from the column.

### Advantages

- Simple.
- Removes a difficult missing-value problem.

### Limitations

- Useful information may be lost.
- The feature may actually be valuable for prediction.

### Interpretation

A column should be removed only after checking its importance and missing-value percentage.

### Insight

High missingness does not automatically mean a feature should be removed.

### AI/ML Relevance

Removing an important predictive feature can reduce model performance.

```python
df_drop_column = df.drop(columns=["Cabin"])

print("Original columns:", df.shape[1])
print("Columns after dropping Cabin:", df_drop_column.shape[1])
```

## 6. Mean Imputation

### Concept

Mean imputation replaces missing numerical values with the average of the available values.

### Example

If the average age is 30, missing age values can be replaced with 30.

### When to Use

- Numerical data.
- Data with a roughly symmetric distribution.
- When there are no strong outliers.

### When Not to Use

- Highly skewed data.
- Data with many outliers.

### Advantages

- Simple and fast.
- Easy to understand.

### Limitations

- Can reduce natural variation.
- Can be affected by outliers.
- May distort the distribution.

### Interpretation

Mean imputation is simple, but it should be selected only when the mean represents the data reasonably well.

### Insight

We should check the distribution before choosing mean imputation.

### AI/ML Relevance

A suitable imputation method helps maintain useful information while making the dataset usable for Machine Learning.

```python
df_mean = df.copy()

df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

print("Missing Age after mean imputation:",
      df_mean["Age"].isnull().sum())
```

## 7. Median Imputation

### Concept

Median imputation replaces missing numerical values with the middle value of the available observations.

### Example

If the median age is 28, missing age values are replaced with 28.

### When to Use

- Numerical data.
- Skewed distributions.
- Data containing outliers.

### When Not to Use

- When the median does not represent the underlying data.
- When missingness requires a more advanced method.

### Advantages

- Less affected by outliers than the mean.
- Simple and robust.

### Limitations

- Reduces variation.
- May not capture relationships between variables.

### Interpretation

For a feature such as `Age`, median imputation can be a practical choice when the distribution is not perfectly symmetric.

### Insight

Median is often safer than mean when numerical data contains skewness or outliers.

### AI/ML Relevance

Good imputation helps prevent missing values from causing errors during model training.

```python
df_median = df.copy()

df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

print("Missing Age after median imputation:",
      df_median["Age"].isnull().sum())
```

## 8. Mode Imputation

### Concept

Mode imputation replaces missing categorical values with the most frequently occurring category.

### Example

If `Embarked` has `S` as the most common value, missing `Embarked` values can be replaced with `S`.

### When to Use

- Categorical variables.
- When one category is clearly dominant.

### When Not to Use

- When categories have similar frequencies.
- When the missingness itself carries useful information.

### Advantages

- Simple and easy to understand.
- Works well for categorical data.

### Limitations

- Can increase the frequency of the most common category.
- May introduce bias.

### Interpretation

Mode is suitable for small amounts of missing categorical data when the most frequent category is a reasonable replacement.

### Insight

Categorical variables should generally not be filled with numerical mean values.

### AI/ML Relevance

Correct categorical imputation prepares the feature for later encoding and model training.

```python
df_mode = df.copy()

df_mode["Embarked"] = df_mode["Embarked"].fillna(
    df_mode["Embarked"].mode()[0]
)

print("Missing Embarked after mode imputation:",
      df_mode["Embarked"].isnull().sum())
```

## 9. Constant Value Imputation

### Concept

Constant imputation replaces missing values with a fixed value such as `"Unknown"` or `0`.

### Example

Missing `Cabin` values can be represented as `"Unknown"` instead of removing the column.

### When to Use

- When missingness itself is meaningful.
- For categorical columns where `"Unknown"` is a valid category.

### When Not to Use

- When the constant value can be confused with a real value.
- When it creates an unrealistic distribution.

### Advantages

- Simple.
- Preserves all rows.

### Limitations

- Creates an artificial category or value.
- May introduce bias.

### Interpretation

A constant such as `"Unknown"` can clearly indicate that the original information was unavailable.

### Insight

Sometimes it is better to preserve missingness as information rather than replacing it with an estimated value.

### AI/ML Relevance

A clear missing category can be encoded later and used by Machine Learning models.

```python
df_constant = df.copy()

df_constant["Cabin"] = df_constant["Cabin"].fillna("Unknown")

print("Missing Cabin after constant imputation:",
      df_constant["Cabin"].isnull().sum())
```

## 10. Forward Fill and Backward Fill

### Concept

Forward fill (`ffill`) uses the previous available value to fill a missing value.

Backward fill (`bfill`) uses the next available value.

### Example

These techniques are commonly useful for time-ordered data such as sensor readings or daily measurements.

### When to Use

- Ordered or time-series data.
- When nearby observations are expected to be similar.

### When Not to Use

- Randomly ordered records.
- Data where neighboring observations have no meaningful relationship.

### Advantages

- Simple.
- Preserves the sequence of the data.

### Limitations

- Can carry an incorrect value forward or backward.
- Not suitable for every dataset.

### Interpretation

For the Titanic dataset, these methods are demonstrated for learning purposes, but they are not automatically the best choice because passenger rows are not a time series.

### AI/ML Relevance

Using an inappropriate fill method can introduce false patterns into the training data.

```python
df_ffill = df.copy()
df_bfill = df.copy()

df_ffill["Age"] = df_ffill["Age"].ffill()
df_bfill["Age"] = df_bfill["Age"].bfill()

print("Remaining Age missing values:")
print("Forward Fill:", df_ffill["Age"].isnull().sum())
print("Backward Fill:", df_bfill["Age"].isnull().sum())
```

## 11. Interpolation

### Concept

Interpolation estimates a missing numerical value using nearby known values.

### Example

If a numerical sequence contains a missing value between two known values, interpolation can estimate a reasonable value between them.

### When to Use

- Ordered numerical data.
- Time-series or continuous measurements.

### When Not to Use

- Unordered categorical data.
- When neighboring values do not have a meaningful relationship.

### Advantages

- Produces a smooth estimate.
- Uses surrounding information.

### Limitations

- Depends on the order of the data.
- May produce unrealistic values when the data is not continuous.

### Interpretation

Interpolation is useful mainly when the position of observations has meaning.

### Insight

The technique should be selected based on the structure of the dataset, not simply because it is available.

### AI/ML Relevance

Appropriate interpolation can preserve trends without unnecessarily deleting observations.

```python
df_interpolated = df.copy()

df_interpolated["Age"] = df_interpolated["Age"].interpolate()

print("Missing Age after interpolation:",
      df_interpolated["Age"].isnull().sum())
```

## 12. Group-Based Imputation

### Concept

Group-based imputation fills missing values using statistics calculated within meaningful groups.

### Example

Instead of using one median age for all passengers, we can calculate the median age separately for each passenger class.

### When to Use

- When groups have meaningful differences.
- When another feature is strongly related to the missing feature.

### When Not to Use

- When groups are too small.
- When the grouping variable has no meaningful relationship.

### Advantages

- More specific than global imputation.
- Can preserve differences between groups.

### Limitations

- Small groups may produce unreliable estimates.
- Requires a meaningful grouping variable.

### Interpretation

Passenger class may provide useful context when estimating missing age values.

### AI/ML Relevance

Group-based imputation can preserve relationships between features and potentially improve the quality of model inputs.

```python
df_group = df.copy()

df_group["Age"] = df_group["Age"].fillna(
    df_group.groupby("Pclass")["Age"].transform("median")
)

print("Missing Age after group-based imputation:",
      df_group["Age"].isnull().sum())
```

## 13. KNN Imputation

### Concept

KNN Imputation estimates missing values using similar observations, called nearest neighbors.

### Example

For a passenger with a missing `Age`, the method can look at similar passengers based on available numerical features.

### When to Use

- Numerical data.
- When similar observations are expected to have similar values.

### When Not to Use

- Very large datasets where computation becomes expensive.
- Data with poor feature scaling or weak similarity information.

### Advantages

- Uses information from multiple features.
- Can capture relationships between observations.

### Limitations

- More computationally expensive.
- Sensitive to feature scale.
- Requires careful selection of features.

### Interpretation

KNN can provide a more data-driven estimate than simple mean or median imputation.

### AI/ML Relevance

It can preserve relationships between features, but the imputation process must be designed carefully to avoid data leakage.

```python
from sklearn.impute import KNNImputer

df_knn = df[["Age", "Fare", "Pclass", "SibSp", "Parch"]].copy()

imputer = KNNImputer(n_neighbors=5)

df_knn_imputed = pd.DataFrame(
    imputer.fit_transform(df_knn),
    columns=df_knn.columns
)

print("Missing values after KNN Imputation:")
print(df_knn_imputed.isnull().sum())
```

## 14. Iterative Imputation

### Concept

Iterative Imputation estimates missing values by modeling each incomplete feature using the other available features.

### Example

Missing `Age` values can be estimated using information such as `Fare`, `Pclass`, `SibSp`, and `Parch`.

### When to Use

- When features have meaningful relationships.
- When a more advanced imputation method is justified.

### When Not to Use

- Very small datasets without enough information.
- When a simple method already provides a reliable solution.

### Advantages

- Uses relationships between multiple variables.
- Can produce more informed estimates.

### Limitations

- More complex.
- More computationally expensive.
- Results depend on the selected model and features.

### Interpretation

Iterative imputation can be useful when several variables can help predict the missing value.

### AI/ML Relevance

It can preserve relationships between features, but it must be applied carefully to avoid information leakage.

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df_iterative = df[["Age", "Fare", "Pclass", "SibSp", "Parch"]].copy()

imputer = IterativeImputer(random_state=42)

df_iterative_imputed = pd.DataFrame(
    imputer.fit_transform(df_iterative),
    columns=df_iterative.columns
)

print("Missing values after Iterative Imputation:")
print(df_iterative_imputed.isnull().sum())
```

## 15. Choosing the Appropriate Technique

Different missing-value techniques are suitable for different situations.

| Technique | Best Used For |
|---|---|
| Drop Rows | Very few missing records |
| Drop Columns | Extremely high missingness or low-value feature |
| Mean | Symmetric numerical data |
| Median | Skewed numerical data or outliers |
| Mode | Categorical data |
| Constant | Meaningful missing category |
| Forward Fill | Ordered or time-series data |
| Backward Fill | Ordered or time-series data |
| Interpolation | Continuous ordered numerical data |
| Group-Based | Meaningful groups |
| KNN | Similar numerical observations |
| Iterative | Strong relationships between features |

### Important

We should not fill every missing numerical value with the mean.

The technique should be selected based on the data type, distribution, amount of missingness, relationships between variables, and the reason for missingness.

```python
comparison = pd.DataFrame({
    "Technique": [
        "Mean",
        "Median",
        "Mode",
        "Constant",
        "Group-Based"
    ],
    "Example Column": [
        "Age",
        "Age",
        "Embarked",
        "Cabin",
        "Age by Pclass"
    ]
})

print(comparison)
```

## 16. Preprocessing Decision

### Problem

The Titanic dataset contains missing values, especially in `Age`, `Cabin`, and `Embarked`.

### Analysis

The amount and nature of missing data are different for each column. Therefore, using one method for every column would not be appropriate.

### Technique Selected

For demonstration, different techniques were tested based on the type and characteristics of each feature.

- `Age` → Median or group-based imputation can be considered.
- `Embarked` → Mode imputation can be considered.
- `Cabin` → Constant value such as `Unknown` or feature removal can be considered depending on the modeling objective.

### Reason

Median is less affected by outliers than mean, mode is suitable for categorical data, and `Unknown` preserves the fact that cabin information was unavailable.

### Implementation

The techniques were tested using separate copies of the dataset so that the original dataset was not unnecessarily modified.

### Result

The selected techniques reduce or remove missing values while preserving as much useful information as possible.

### Impact

Proper missing-value handling allows Machine Learning algorithms to work with complete input data and can reduce bias caused by inappropriate deletion or imputation.

### Important

These decisions identify possible preprocessing actions. The complete final preprocessing should be performed in the appropriate preprocessing stage rather than automatically applying every technique here.

```python
print("Original Missing Values:")
print(df.isnull().sum())

print("\nRecommended Initial Treatment:")
print("Age      -> Median / Group-Based Imputation")
print("Embarked -> Mode Imputation")
print("Cabin    -> Unknown category / Drop after further analysis")
```
