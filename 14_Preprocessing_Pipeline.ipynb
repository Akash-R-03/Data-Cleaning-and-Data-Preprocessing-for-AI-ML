# 14 – Scikit-learn Preprocessing Pipeline

## Objective

This notebook explains how to build a complete preprocessing pipeline using Scikit-learn.

The main workflow is:

**Numerical Features → Imputation → Scaling**

**Categorical Features → Imputation → Encoding**

The notebook also explains `Pipeline`, `ColumnTransformer`, `fit()`, `transform()`, and `fit_transform()`.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 1. What is a Preprocessing Pipeline?

A preprocessing pipeline is a sequence of data preprocessing steps performed in a fixed order.

For example:

Numerical Features:
**Missing Value Imputation → Scaling**

Categorical Features:
**Missing Value Imputation → One-Hot Encoding**

### Why use Pipelines?

Pipelines help to:

- Organize preprocessing steps.
- Apply transformations consistently.
- Reduce manual coding.
- Prevent data leakage.
- Ensure the same preprocessing is applied to training and test data.
- Make Machine Learning workflows easier to reproduce.

### Required Documentation

**Problem:**  
Different preprocessing steps need to be applied to different types of features.

**Analysis:**  
Numerical and categorical features require different preprocessing techniques.

**Technique Selected:**  
Scikit-learn `Pipeline` and `ColumnTransformer`.

**Reason:**  
They allow multiple preprocessing steps to be combined into one structured workflow.

**Implementation:**  
Separate numerical and categorical pipelines are created and combined.

**Result:**  
The preprocessing workflow becomes organized and reusable.

**Impact:**  
A consistent preprocessing process can improve model reliability and reduce data leakage.

```python
print("Numerical and categorical preprocessing will be handled using separate pipelines.")
```

## 2. Why Use Pipelines?

A pipeline combines multiple preprocessing operations into a single workflow.

Instead of manually performing:

1. Missing value handling
2. Scaling
3. Encoding

we can place these operations inside a pipeline.

This ensures that the steps are executed in the correct order.

### Advantages

- Simple and organized workflow.
- Prevents inconsistent preprocessing.
- Helps prevent data leakage.
- Easy to connect with Machine Learning models.
- Makes the workflow reproducible.

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

print("Numerical Pipeline:")
print(numeric_pipeline)

print("\nCategorical Pipeline:")
print(categorical_pipeline)
```

## 3. What is ColumnTransformer?

`ColumnTransformer` allows different preprocessing pipelines to be applied to different columns of the dataset.

For example:

- Numerical columns → Median Imputation → Standard Scaling
- Categorical columns → Most-Frequent Imputation → One-Hot Encoding

This is useful because a dataset usually contains different types of features.

### Required Documentation

**Problem:**  
Numerical and categorical columns cannot always use the same preprocessing method.

**Analysis:**  
Numerical features require numerical transformations, while categorical features require categorical transformations.

**Technique Selected:**  
`ColumnTransformer`.

**Reason:**  
It applies the appropriate preprocessing pipeline to the correct columns.

**Implementation:**  
Numerical and categorical pipelines are passed to `ColumnTransformer`.

**Result:**  
Each feature type receives suitable preprocessing.

**Impact:**  
This produces a consistent feature matrix suitable for Machine Learning.

```python
numeric_features = ["Age", "SibSp", "Parch", "Fare"]
categorical_features = ["Pclass", "Sex", "Embarked"]

preprocessor = ColumnTransformer([
    ("numerical", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessor created successfully.")
```

## 4. Numerical Pipeline – Imputation and Scaling

Numerical features may contain missing values and may have different scales.

The numerical preprocessing workflow is:

**Numerical Features → Median Imputation → Standard Scaling**

### Median Imputation

Median is used to replace missing numerical values.

It is useful when numerical data may contain outliers because the median is less affected by extreme values than the mean.

### StandardScaler

`StandardScaler` transforms numerical values so that they are approximately centered around 0 with a standard deviation of 1.

### When to use?

This approach is useful for numerical features when missing values exist and feature scale may affect the Machine Learning algorithm.

### Required Documentation

**Problem:**  
Numerical columns may contain missing values and different value ranges.

**Analysis:**  
Missing values can prevent some algorithms from working, while different scales can affect scale-sensitive algorithms.

**Technique Selected:**  
Median imputation followed by `StandardScaler`.

**Reason:**  
Median imputation is relatively robust to outliers, while standardization puts numerical features on a comparable scale.

**Implementation:**  
`SimpleImputer(strategy="median")` followed by `StandardScaler()`.

**Result:**  
Missing numerical values are replaced and numerical features are standardized.

**Impact:**  
This can improve the performance of scale-sensitive Machine Learning algorithms.

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

print("Numerical pipeline:")
print(numeric_pipeline)
```

## 5. Categorical Pipeline – Imputation and Encoding

Categorical features can contain missing values and text-based categories.

The categorical preprocessing workflow is:

**Categorical Features → Imputation → One-Hot Encoding**

### Most-Frequent Imputation

Missing categorical values are replaced with the most frequently occurring category.

### One-Hot Encoding

One-hot encoding converts categorical values into numerical binary columns.

For example:

Sex:

- male
- female

can become separate numerical columns.

`handle_unknown="ignore"` allows the pipeline to handle categories that were not present during training.

### Required Documentation

**Problem:**  
Categorical features may contain missing values and text categories that Machine Learning algorithms cannot directly process.

**Analysis:**  
Missing categories need to be handled, and categorical values need to be converted into numerical representation.

**Technique Selected:**  
Most-frequent imputation followed by One-Hot Encoding.

**Reason:**  
The most frequent category provides a simple categorical imputation method, while One-Hot Encoding avoids creating an artificial numerical order.

**Implementation:**  
`SimpleImputer(strategy="most_frequent")` followed by `OneHotEncoder(handle_unknown="ignore")`.

**Result:**  
Missing categorical values are replaced and categories are converted into numerical features.

**Impact:**  
The transformed categorical features can be processed by Machine Learning algorithms.

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

print("Categorical pipeline:")
print(categorical_pipeline)
```

## 6. Combining Multiple Transformations

`ColumnTransformer` combines the numerical and categorical pipelines.

The final preprocessing workflow is:

### Numerical Features
**Age, SibSp, Parch, Fare**

→ Median Imputation

→ Standard Scaling

### Categorical Features
**Pclass, Sex, Embarked**

→ Most-Frequent Imputation

→ One-Hot Encoding

This allows different preprocessing methods to be applied to different feature types within the same dataset.

```python
numeric_features = ["Age", "SibSp", "Parch", "Fare"]
categorical_features = ["Pclass", "Sex", "Embarked"]

preprocessor = ColumnTransformer([
    ("numerical", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Complete preprocessing structure:")
print(preprocessor)
```

## 7. Preparing Features and Target

The target variable is separated from the input features before preprocessing.

For this Titanic dataset:

- Target: `Survived`
- Features: Selected numerical and categorical columns

The target should not be included in the preprocessing pipeline because it is the value the model is expected to predict.

```python
X = df[numeric_features + categorical_features].copy()
y = df["Survived"].copy()

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

X.head()
```

## 8. Train-Test Split

The dataset is divided into training and testing data before fitting the preprocessing pipeline.

This is important because preprocessing should be learned only from the training data.

### Why?

If information from the test dataset is used while fitting preprocessing steps, data leakage can occur.

The test dataset should represent unseen data.

A fixed `random_state` makes the split reproducible.

`stratify=y` maintains a similar target-class distribution in both training and testing datasets.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

## 9. Understanding fit()

`fit()` learns the required information from the training data.

For example:

- `SimpleImputer` learns the median or most frequent value.
- `StandardScaler` learns the mean and standard deviation.
- `OneHotEncoder` learns the categories.

The learned information is obtained only from the training dataset.

### Important Rule

**Fit preprocessing only on training data.**

Do not fit the preprocessing pipeline using the test dataset.

```python
preprocessor.fit(X_train)

print("Preprocessor fitted using training data.")
```

## 10. Understanding transform()

`transform()` applies the preprocessing rules that were already learned during `fit()`.

The test data is transformed using the information learned from the training data.

This prevents information from the test dataset from influencing the preprocessing process.

```python
X_train_transformed = preprocessor.transform(X_train)
X_test_transformed = preprocessor.transform(X_test)

print("Transformed Training Shape:", X_train_transformed.shape)
print("Transformed Testing Shape:", X_test_transformed.shape)
```

## 11. Understanding fit_transform()

`fit_transform()` performs two operations:

1. `fit()` – learns preprocessing information.
2. `transform()` – applies the learned preprocessing.

For training data, `fit_transform()` is commonly used.

For test data, only `transform()` should be used.

### Correct Workflow

**Training Data → fit_transform()**

**Testing Data → transform()**

### Incorrect Workflow

**Training Data → fit_transform()**

**Testing Data → fit_transform()**

The incorrect workflow can cause data leakage because the preprocessing would learn information from the test dataset.

```python
X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed = preprocessor.transform(X_test)

print("Training data transformed using fit_transform().")
print("Testing data transformed using transform().")

print("\nTraining Shape:", X_train_transformed.shape)
print("Testing Shape:", X_test_transformed.shape)
```

## 12. Complete Scikit-learn Preprocessing Pipeline

A complete preprocessing pipeline can combine the `ColumnTransformer` with a Machine Learning model.

The complete workflow becomes:

**Raw Data**

↓

**Train-Test Split**

↓

**Numerical Features → Imputation → Scaling**

↓

**Categorical Features → Imputation → Encoding**

↓

**Machine Learning Model**

This structure ensures that preprocessing is performed consistently when the model is trained and evaluated.

```python
from sklearn.linear_model import LogisticRegression

model_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

model_pipeline.fit(X_train, y_train)

accuracy = model_pipeline.score(X_test, y_test)

print("Test Accuracy:", accuracy)
```

## 13. Checking the Transformed Feature Names

After preprocessing, the original columns may become multiple numerical columns.

For example, categorical columns processed with One-Hot Encoding can produce separate columns for their categories.

Checking feature names helps us understand what the Machine Learning model receives after preprocessing.

```python
feature_names = preprocessor.get_feature_names_out()

print("Number of transformed features:", len(feature_names))
print("\nTransformed Feature Names:")
print(feature_names)
```

```python
print("Preprocessing Pipeline Completed Successfully")

print("\nWorkflow:")
print("1. Train-Test Split")
print("2. Numerical Features -> Median Imputation -> Standard Scaling")
print("3. Categorical Features -> Most-Frequent Imputation -> One-Hot Encoding")
print("4. Combine using ColumnTransformer")
print("5. Train Machine Learning Model")
print("6. Evaluate on Test Data")

print("\nFinal Test Accuracy:", model_pipeline.score(X_test, y_test))
```
