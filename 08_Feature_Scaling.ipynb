# 1. Why Feature Scaling?

### Question: Why is Feature Scaling important?

Feature scaling is the process of transforming numerical features so that they are on a similar scale.

In the Titanic dataset, numerical columns such as `Age`, `Fare`, `SibSp`, `Parch`, and `Pclass` have different ranges.

### Insight

Features with very different ranges can affect scale-sensitive Machine Learning algorithms differently.

### AI/ML Relevance

Feature scaling helps scale-sensitive algorithms work more effectively by reducing the influence of differences in feature magnitude.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)

numeric_cols = df.select_dtypes(include="number").columns

print("\nNumerical Columns:")
print(list(numeric_cols))
```

# 2. Scale-Sensitive Algorithms

### Question: What are Scale-Sensitive Algorithms?

Scale-sensitive algorithms are Machine Learning algorithms whose performance can be affected by differences in the scale of numerical features.

Examples include:

- K-Nearest Neighbors (KNN)
- Support Vector Machine (SVM)
- Logistic Regression
- Linear Regression
- K-Means Clustering
- Principal Component Analysis (PCA)

### Insight

These algorithms use distances, magnitudes, or mathematical calculations where differences in feature scales can influence the result.

### AI/ML Relevance

Scaling numerical features can help these algorithms treat features more fairly.

```python
print("Scale-Sensitive Algorithms:")
print([
    "K-Nearest Neighbors (KNN)",
    "Support Vector Machine (SVM)",
    "Logistic Regression",
    "Linear Regression",
    "K-Means Clustering",
    "Principal Component Analysis (PCA)"
])
```

# 3. Scale-Insensitive Algorithms

### Question: What are Scale-Insensitive Algorithms?

Scale-insensitive algorithms are algorithms that generally do not require feature scaling because their performance is not strongly affected by differences in feature magnitude.

Examples include:

- Decision Tree
- Random Forest
- Gradient Boosting

### Insight

Tree-based algorithms split data using conditions on individual features rather than relying mainly on distances between observations.

### AI/ML Relevance

Feature scaling is generally not necessary for tree-based algorithms.

```python
print("Scale-Insensitive Algorithms:")
print([
    "Decision Tree",
    "Random Forest",
    "Gradient Boosting"
])
```

# 4. Normalization

### Question: What is Normalization?

Normalization transforms numerical values into a specific range.

A common normalization formula is:

X_scaled = (X - X_min) / (X_max - X_min)

The result is commonly between 0 and 1.

### When to use

Normalization is useful when features need to be represented within a fixed range.

### When not to use

It may not be suitable when strong outliers are present because minimum and maximum values can be affected by extreme observations.

### Advantages

- Produces a fixed range.
- Easy to understand.
- Useful for distance-based algorithms.

### Limitations

- Sensitive to outliers.
- Extreme values can compress the remaining observations.

### AI/ML Relevance

Normalization can improve the performance of algorithms that are sensitive to feature magnitude.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(df[numeric_cols])

normalized_df = pd.DataFrame(
    normalized_data,
    columns=numeric_cols
)

print(normalized_df.head())
```

# 5. Standardization

### Question: What is Standardization?

Standardization transforms numerical features so that they have a mean close to 0 and a standard deviation close to 1.

The formula is:

X_scaled = (X - mean) / standard_deviation

### When to use

Standardization is commonly used for scale-sensitive algorithms such as Logistic Regression, SVM, KNN, and PCA.

### When not to use

It may not be suitable when extreme outliers strongly affect the mean and standard deviation.

### Advantages

- Centers data around zero.
- Makes feature scales more comparable.
- Commonly useful for many statistical and Machine Learning algorithms.

### Limitations

- Sensitive to outliers.
- Does not produce a fixed range.

### AI/ML Relevance

Standardization can help scale-sensitive algorithms perform more consistently.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

standardized_data = scaler.fit_transform(df[numeric_cols])

standardized_df = pd.DataFrame(
    standardized_data,
    columns=numeric_cols
)

print(standardized_df.head())
```

# 6. Min-Max Scaling

### Question: What is Min-Max Scaling?

Min-Max Scaling transforms numerical values into a specified range, commonly 0 to 1.

The formula is:

X_scaled = (X - X_min) / (X_max - X_min)

### When to use

Use Min-Max Scaling when a bounded range is useful and the data does not contain problematic extreme outliers.

### When not to use

Avoid relying on Min-Max Scaling when strong outliers can distort the minimum and maximum values.

### Advantages

- Produces a fixed range.
- Simple to understand.
- Maintains the relative ordering of values.

### Limitations

- Sensitive to outliers.
- Extreme values can affect the scaling of other observations.

### AI/ML Relevance

Min-Max Scaling is useful for algorithms where having features in a common range is beneficial.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

minmax_data = scaler.fit_transform(df[numeric_cols])

minmax_df = pd.DataFrame(
    minmax_data,
    columns=numeric_cols
)

print(minmax_df.head())
```

# 7. StandardScaler

### Question: What is StandardScaler?

`StandardScaler` standardizes numerical features by removing the mean and scaling the data according to its standard deviation.

### When to use

It is useful for scale-sensitive algorithms such as:

- Logistic Regression
- SVM
- KNN
- PCA

### When not to use

It should be used carefully when the data contains strong outliers because the mean and standard deviation can be affected.

### Advantages

- Centers features around zero.
- Makes feature scales comparable.
- Commonly used in Machine Learning preprocessing.

### Limitations

- Sensitive to outliers.
- Does not restrict values to a fixed range.

### AI/ML Relevance

StandardScaler is useful when a Machine Learning algorithm performs better with standardized numerical features.

```python
scaler = StandardScaler()

scaled_data = scaler.fit_transform(df[numeric_cols])

scaled_df = pd.DataFrame(
    scaled_data,
    columns=numeric_cols
)

print("StandardScaler Result:")
print(scaled_df.head())
```

# 8. MinMaxScaler

### Question: What is MinMaxScaler?

`MinMaxScaler` transforms numerical features into a selected range, normally between 0 and 1.

### When to use

It is useful when numerical features need to have a common bounded range.

### When not to use

It may not be appropriate when strong outliers are present.

### Advantages

- Produces values in a fixed range.
- Easy to interpret.
- Preserves the relative ordering of values.

### Limitations

- Sensitive to outliers.
- Minimum and maximum values influence the transformation.

### AI/ML Relevance

MinMaxScaler can be useful for scale-sensitive algorithms that benefit from features being within a common range.

```python
scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df[numeric_cols])

minmax_scaled_df = pd.DataFrame(
    scaled_data,
    columns=numeric_cols
)

print("MinMaxScaler Result:")
print(minmax_scaled_df.head())
```

# 9. RobustScaler

### Question: What is RobustScaler?

`RobustScaler` scales numerical features using the median and interquartile range (IQR).

It is less affected by extreme values than methods based on the mean and standard deviation.

### When to use

Use RobustScaler when numerical features contain significant outliers.

### When not to use

It may not be necessary when the data has few or no problematic outliers.

### Advantages

- More resistant to outliers.
- Uses median and IQR.
- Useful for skewed numerical data.

### Limitations

- Does not produce a fixed range.
- The transformed values may be less intuitive to interpret.

### AI/ML Relevance

RobustScaler can reduce the influence of outliers when scaling numerical features for scale-sensitive algorithms.

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

robust_data = scaler.fit_transform(df[numeric_cols])

robust_df = pd.DataFrame(
    robust_data,
    columns=numeric_cols
)

print("RobustScaler Result:")
print(robust_df.head())
```

# 10. MaxAbsScaler

### Question: What is MaxAbsScaler?

`MaxAbsScaler` scales each numerical feature using its maximum absolute value.

The transformed values are generally within the range -1 to 1.

### When to use

It is useful when the original zero value should remain unchanged and when preserving sparsity is important.

### When not to use

It may not be suitable when extreme maximum values strongly affect the scaling.

### Advantages

- Does not shift the data.
- Preserves zero values.
- Useful for sparse datasets.

### Limitations

- Sensitive to extreme values.
- Does not remove the effect of outliers.

### AI/ML Relevance

MaxAbsScaler can be useful when scaling is required while preserving the structure of sparse data.

```python
from sklearn.preprocessing import MaxAbsScaler

scaler = MaxAbsScaler()

maxabs_data = scaler.fit_transform(df[numeric_cols])

maxabs_df = pd.DataFrame(
    maxabs_data,
    columns=numeric_cols
)

print("MaxAbsScaler Result:")
print(maxabs_df.head())
```

# 11. Compare Different Scaling Techniques

### Question: How do different scaling techniques compare?

Different scaling techniques have different purposes and different levels of sensitivity to outliers.

| Technique | Main Idea | Outlier Sensitivity | Typical Range |
|---|---|---|---|
| StandardScaler | Mean and Standard Deviation | High | No fixed range |
| MinMaxScaler | Minimum and Maximum | High | 0 to 1 |
| RobustScaler | Median and IQR | Lower | No fixed range |
| MaxAbsScaler | Maximum Absolute Value | High | -1 to 1 |

### Insight

There is no single scaling technique that is best for every dataset.

### AI/ML Relevance

The scaling technique should be selected based on the feature distribution, presence of outliers, and requirements of the Machine Learning algorithm.

```python
comparison = pd.DataFrame({
    "Original Fare": df["Fare"].head(10).values,
    "StandardScaler": standardized_df["Fare"].head(10).values,
    "MinMaxScaler": minmax_df["Fare"].head(10).values,
    "RobustScaler": robust_df["Fare"].head(10).values,
    "MaxAbsScaler": maxabs_df["Fare"].head(10).values
})

print(comparison)
```

# 12. How Outliers Affect Scaling

### Question: How do outliers affect scaling?

Outliers can strongly influence scaling techniques that depend on the minimum, maximum, mean, or standard deviation.

For example, the `Fare` column in the Titanic dataset contains some very high values.

When Min-Max Scaling is applied, these extreme values can affect the minimum and maximum range and compress many normal observations into a smaller part of the scaled range.

### Insight

An outlier should not automatically be removed just because it affects scaling. It may represent a valid extreme observation.

### AI/ML Relevance

Understanding the effect of outliers helps us select an appropriate scaling technique for Machine Learning.

```python
print("Fare Statistics:")
print(df["Fare"].describe())

print("\nLargest Fare Values:")
print(df["Fare"].sort_values(ascending=False).head(10))
```
