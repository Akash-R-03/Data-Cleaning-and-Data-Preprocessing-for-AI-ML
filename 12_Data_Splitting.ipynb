# Notebook 12 – Train, Validation & Test Data

## 1. What is a Training Dataset?

A training dataset is the portion of data used to train a Machine Learning model.

The model learns patterns and relationships from the training data.

### Example

If we have 1000 records, we may use 700 records for training.

### Why is it important?

The training dataset provides the information required for the model to learn from historical examples.

### AI/ML Relevance

A model learns its parameters from training data. The quality and representativeness of the training data directly affect model performance.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())
```

## 2. What is a Validation Dataset?

A validation dataset is used during model development to evaluate different models, hyperparameters, or configurations.

It helps us select the best model before evaluating it on the final test dataset.

### Example

For 1000 records:

- Training: 700 records
- Validation: 150 records
- Test: 150 records

### AI/ML Relevance

Validation data helps prevent selecting a model that performs well only on the training data.

```python
from sklearn.model_selection import train_test_split

train_data, temp_data = train_test_split(
    df,
    test_size=0.30,
    random_state=42
)

validation_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42
)

print("Training Data:", train_data.shape)
print("Validation Data:", validation_data.shape)
print("Test Data:", test_data.shape)
```

## 3. What is a Test Dataset?

A test dataset is a separate portion of data used for the final evaluation of a Machine Learning model.

The test data should not be used for training or model selection.

### Why is it important?

It provides an estimate of how the final model may perform on unseen data.

### AI/ML Relevance

Keeping test data separate helps provide a more realistic evaluation of model generalization.

```python
print("Training Records:", len(train_data))
print("Validation Records:", len(validation_data))
print("Test Records:", len(test_data))

print("\nTest Dataset Preview:")
print(test_data.head())
```

## 4. What is Train-Test Split?

Train-Test Split divides a dataset into two parts:

- Training data – used to train the model.
- Test data – used to evaluate the final model.

A common split is 80% training and 20% testing.

### AI/ML Relevance

Train-test splitting allows us to evaluate whether a model can generalize to unseen data.

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)

print("Training Shape:", train.shape)
print("Testing Shape:", test.shape)
```

## 5. How to Perform Train-Validation-Test Split?

A Train-Validation-Test split divides the dataset into three independent parts.

- Training data: Used to train the model.
- Validation data: Used for model selection and tuning.
- Test data: Used for final evaluation.

A common approach is:

- 70% Training
- 15% Validation
- 15% Testing

### AI/ML Relevance

This separation prevents the test dataset from influencing model development.

```python
train_data, temp_data = train_test_split(
    df,
    test_size=0.30,
    random_state=42
)

validation_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42
)

print("Training:", len(train_data))
print("Validation:", len(validation_data))
print("Testing:", len(test_data))
```

## 6. What is Random State?

`random_state` controls the randomness used during operations such as train-test splitting.

Using the same `random_state` produces the same split each time.

### Why is it important?

It makes experiments reproducible.

### AI/ML Relevance

Reproducibility allows developers and researchers to obtain consistent results when testing and comparing Machine Learning models.

```python
train1, test1 = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)

train2, test2 = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)

print("Same training split:", train1.index.equals(train2.index))
print("Same testing split:", test1.index.equals(test2.index))
```

## 7. What is Stratified Sampling?

Stratified sampling ensures that the proportion of each target class is approximately maintained in the training and testing datasets.

This is especially useful for classification problems with imbalanced classes.

### Example

If the original dataset contains:

- 60% Class 0
- 40% Class 1

Stratified splitting attempts to maintain a similar proportion in both datasets.

### AI/ML Relevance

Stratification helps prevent important classes from being underrepresented in the training or testing data.

```python
train_strat, test_strat = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["Survived"]
)

print("Original Class Distribution:")
print(df["Survived"].value_counts(normalize=True))

print("\nTraining Class Distribution:")
print(train_strat["Survived"].value_counts(normalize=True))

print("\nTesting Class Distribution:")
print(test_strat["Survived"].value_counts(normalize=True))
```

## 8. What is Data Leakage?

Data leakage occurs when information from outside the training data is used during model training.

This can happen when information from the validation or test dataset influences preprocessing, feature engineering, or model training.

### Why is it dangerous?

Data leakage can produce unrealistically high model performance during evaluation.

### AI/ML Impact

A model affected by leakage may perform well during development but perform poorly on genuinely unseen data.

```python
print("Example of information that must remain separate:")
print("- Training data")
print("- Validation data")
print("- Test data")

print("\nTest data should not be used while training or selecting the model.")
```

## 9. What is Temporal Splitting?

Temporal splitting is used when data has a time component.

Instead of randomly splitting the data, we train on earlier observations and test on later observations.

### Example

Historical data:

2021 → 2022 → 2023 → 2024

Possible split:

- Training: 2021–2023
- Testing: 2024

### Why not use random splitting?

Random splitting can allow future information to appear in the training data.

### AI/ML Relevance

Temporal splitting is important for forecasting, financial data, sales prediction, and other time-dependent problems.

```python
# Example temporal dataset

time_data = pd.DataFrame({
    "Year": [2021, 2021, 2022, 2022, 2023, 2023, 2024, 2024],
    "Sales": [100, 110, 120, 130, 140, 150, 160, 170]
})

train_time = time_data[time_data["Year"] <= 2023]
test_time = time_data[time_data["Year"] > 2023]

print("Training Data:")
print(train_time)

print("\nTesting Data:")
print(test_time)
```

## 10. Why Must Preprocessing Be Fitted Only on Training Data?

Preprocessing techniques such as scaling, imputation, and encoding may learn information from the data.

Therefore, the preprocessing object should be fitted only using the training dataset.

The same fitted transformation is then applied to the validation and test datasets.

### Incorrect Workflow

1. Combine all data.
2. Fit the scaler using the complete dataset.
3. Split the data.

This allows information from the test data to influence preprocessing.

### Correct Workflow

1. Split the data.
2. Fit preprocessing only on training data.
3. Transform training data.
4. Transform validation and test data using the same fitted preprocessing object.

### AI/ML Impact

The correct workflow prevents data leakage and provides a more reliable estimate of model performance.

```python
from sklearn.preprocessing import StandardScaler

features = ["Age", "Fare"]

data = df[features].copy()
data = data.fillna(data.median())

X_train, X_test = train_test_split(
    data,
    test_size=0.20,
    random_state=42
)

scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Use the same fitted scaler on test data
X_test_scaled = scaler.transform(X_test)

print("Training data after scaling:")
print(X_train_scaled[:5])

print("\nTest data after scaling:")
print(X_test_scaled[:5])
```

## 11. Demonstrate an Incorrect Preprocessing Workflow

### Problem

The scaler is fitted using the complete dataset before splitting.

### Analysis

The scaler learns statistics such as the mean and standard deviation from both training and test data.

Therefore, information from the test dataset indirectly influences the training process.

### Technique Selected

StandardScaler is used to demonstrate the preprocessing workflow.

### Reason

StandardScaler is commonly used for numerical features and clearly demonstrates why preprocessing must be fitted only on training data.

### Implementation

The scaler is fitted on the complete dataset before train-test splitting.

### Result

The scaling parameters contain information from the test data.

### Impact

This creates data leakage and may result in an overly optimistic evaluation.

```python
from sklearn.preprocessing import StandardScaler

scaler_incorrect = StandardScaler()

# Incorrect: fitting on the complete dataset
data_scaled_incorrect = scaler_incorrect.fit_transform(data)

X_train_wrong, X_test_wrong = train_test_split(
    data_scaled_incorrect,
    test_size=0.20,
    random_state=42
)

print("Incorrect workflow completed.")
print("The scaler was fitted using the complete dataset.")
print("This can cause data leakage.")
```

## 12. Demonstrate the Correct Preprocessing Workflow

### Problem

Preprocessing must not use information from validation or test data.

### Analysis

If preprocessing parameters are calculated using the complete dataset, information from unseen data can leak into the training process.

### Technique Selected

StandardScaler with train-test splitting.

### Reason

The scaler is fitted only on the training dataset and then applied to the test dataset.

### Implementation

1. Split the data.
2. Fit the scaler on training data.
3. Transform training data.
4. Transform test data.

### Result

The test data does not influence the scaler's learned parameters.

### Impact

This reduces data leakage and gives a more trustworthy estimate of Machine Learning performance.

```python
from sklearn.preprocessing import StandardScaler

X_train, X_test = train_test_split(
    data,
    test_size=0.20,
    random_state=42
)

scaler_correct = StandardScaler()

# Fit only on training data
X_train_scaled = scaler_correct.fit_transform(X_train)

# Transform test data using the training-fitted scaler
X_test_scaled = scaler_correct.transform(X_test)

print("Correct workflow completed.")
print("Scaler fitted only on training data.")
print("Test data was transformed using the same scaler.")
```
