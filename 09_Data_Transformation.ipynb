# 1. Why Transform Data?

### Question: Why do we transform data?

Data transformation is the process of changing the format or distribution of numerical data to make it more suitable for analysis and Machine Learning.

In the Titanic dataset, some numerical features such as `Fare` may have a skewed distribution. Transformation techniques can help reduce skewness and make the data easier for some Machine Learning algorithms to use.

### Example

If a feature contains many small values and a few very large values, a transformation can reduce the effect of those large values.

### Insight

Data transformation can help make numerical features more balanced and easier to analyze.

### AI/ML Relevance

Properly transformed features can improve the performance of some Machine Learning algorithms, especially algorithms that are sensitive to feature distribution.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)

print("\nNumerical Columns:")
print(list(df.select_dtypes(include="number").columns))
```

# 2. Log Transformation

### Question: What is Log Transformation?

Log transformation applies the logarithm to numerical values.

It is commonly used for positively skewed data, especially when a feature contains some very large values.

### Example

The `Fare` feature contains some very high values compared with most passengers. A log transformation can reduce the influence of these extreme values.

### When to use

Use log transformation for positive and highly right-skewed data.

### When not to use

Standard log transformation should not be directly applied to zero or negative values.

### Advantages

- Reduces right skewness.
- Reduces the influence of very large values.

### Limitation

It is not directly suitable for zero or negative values unless an appropriate adjustment is made.

### AI/ML Relevance

Reducing strong skewness can help some Machine Learning algorithms learn patterns more effectively.

```python
import numpy as np

df["Fare_Log"] = np.log1p(df["Fare"])

print(df[["Fare", "Fare_Log"]].head())
```

# 3. Square Root Transformation

### Question: What is Square Root Transformation?

Square root transformation applies the square root to numerical values.

It is useful for reducing moderate right skewness.

### Example

A numerical feature with many smaller values and fewer larger values can sometimes become more balanced after square root transformation.

### When to use

It can be used for non-negative numerical data with moderate skewness.

### When not to use

It should not be directly applied to negative values.

### Advantages

- Simple to apply.
- Can reduce moderate right skewness.

### Limitation

It may not be strong enough for highly skewed data.

### AI/ML Relevance

Reducing skewness can make some numerical features more suitable for statistical and Machine Learning models.

```python
df["Fare_Sqrt"] = np.sqrt(df["Fare"])

print(df[["Fare", "Fare_Sqrt"]].head())
```

# 4. Power Transformation

### Question: What is Power Transformation?

Power transformation changes the scale and distribution of numerical data using a mathematical power function.

It is used to make data more symmetric and reduce skewness.

### Example

A highly skewed numerical feature can be transformed to produce a distribution that is closer to a normal shape.

### When to use

Use it when numerical data is skewed and a more balanced distribution is required.

### Advantages

- Can reduce skewness.
- Can make data more suitable for some Machine Learning algorithms.

### Limitation

The result depends on the selected transformation method and parameter.

### AI/ML Relevance

Power transformations can improve the behavior of algorithms that work better when numerical features have a more symmetric distribution.

```python
from sklearn.preprocessing import PowerTransformer

fare_data = df[["Fare"]]

power = PowerTransformer(method="yeo-johnson")

fare_power = power.fit_transform(fare_data)

df["Fare_Power"] = fare_power

print(df[["Fare", "Fare_Power"]].head())
```

# 5. Box-Cox Transformation

### Question: What is Box-Cox Transformation?

Box-Cox transformation is a power transformation used to reduce skewness and make positive numerical data closer to a normal distribution.

### When to use

It is used when the numerical feature contains only positive values.

### When not to use

Box-Cox cannot be directly applied to zero or negative values.

### Advantages

- Can reduce skewness.
- Can make a distribution more symmetric.

### Limitation

It requires strictly positive values.

### AI/ML Relevance

A more suitable distribution can improve the behavior of some statistical and Machine Learning algorithms.

```python
from scipy.stats import boxcox

fare_positive = df["Fare"] + 1

fare_boxcox, lambda_value = boxcox(fare_positive)

print("Lambda:", lambda_value)
print("\nFirst 5 transformed values:")
print(fare_boxcox[:5])
```

# 6. Yeo-Johnson Transformation

### Question: What is Yeo-Johnson Transformation?

Yeo-Johnson is a power transformation that can handle both positive and negative values, including zero.

It is useful when the data does not satisfy the positive-value requirement of Box-Cox transformation.

### When to use

Use it when the feature contains zero or negative values or when a more flexible power transformation is required.

### Advantages

- Can handle zero values.
- Can handle negative values.
- Can reduce skewness.

### Limitation

The transformation may not always produce a perfectly normal distribution.

### AI/ML Relevance

Yeo-Johnson is useful when transforming real-world numerical features that contain zero or negative values.

```python
from sklearn.preprocessing import PowerTransformer

yeo_johnson = PowerTransformer(method="yeo-johnson")

fare_yeo = yeo_johnson.fit_transform(df[["Fare"]])

df["Fare_YeoJohnson"] = fare_yeo

print(df[["Fare", "Fare_YeoJohnson"]].head())
```

# 7. Handling Skewed Data

### Question: How do we handle skewed data?

Skewed data can be handled using suitable transformation techniques such as:

- Log Transformation
- Square Root Transformation
- Power Transformation
- Box-Cox Transformation
- Yeo-Johnson Transformation

The technique should be selected based on the feature distribution and the presence of zero or negative values.

### Example

The `Fare` feature in the Titanic dataset can be checked for skewness before selecting a transformation.

### Insight

Transformation should not be applied automatically. First, we should understand the distribution and then select a suitable technique.

### AI/ML Relevance

Reducing strong skewness can improve the performance of algorithms that are sensitive to feature distribution.

```python
print("Fare Skewness Before Transformation:")
print(df["Fare"].skew())

print("\nFare Skewness After Log Transformation:")
print(df["Fare_Log"].skew())

print("\nFare Skewness After Square Root Transformation:")
print(df["Fare_Sqrt"].skew())
```

# 8. Normalization

### Question: What is Normalization?

Normalization transforms numerical values into a common range, commonly between 0 and 1.

A common Min-Max formula is:

X_scaled = (X - X_min) / (X_max - X_min)

### Example

The `Fare` feature can be normalized so that its values fall between 0 and 1.

### When to use

Use normalization when a fixed range is useful, especially for scale-sensitive algorithms.

### Advantage

It creates a common range for numerical features.

### Limitation

It is sensitive to outliers.

### AI/ML Relevance

Normalization can help algorithms that are affected by differences in feature magnitude.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df["Fare_Normalized"] = scaler.fit_transform(df[["Fare"]])

print(df[["Fare", "Fare_Normalized"]].head())
```

# 9. Standardization

### Question: What is Standardization?

Standardization transforms numerical data so that it has a mean close to 0 and a standard deviation close to 1.

The formula is:

X_scaled = (X - mean) / standard_deviation

### Example

The `Fare` feature can be standardized to make its scale more suitable for algorithms such as Logistic Regression, SVM, KNN, and PCA.

### When to use

Use standardization when the Machine Learning algorithm is sensitive to feature scale.

### Advantage

It centers the feature around zero and makes the scale more comparable.

### Limitation

It can be affected by extreme outliers.

### AI/ML Relevance

Standardization is commonly used before applying scale-sensitive Machine Learning algorithms.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df["Fare_Standardized"] = scaler.fit_transform(df[["Fare"]])

print(df[["Fare", "Fare_Standardized"]].head())
```

# 10. Compare Distribution Before and After Transformation

### Question: How does the distribution change before and after transformation?

We can compare the skewness of the original `Fare` feature with its transformed versions.

A lower absolute skewness value generally indicates a more symmetric distribution.

### Insight

The transformation may reduce the skewness of the `Fare` feature, but the best transformation should be selected based on the actual data and Machine Learning requirement.

### AI/ML Relevance

Comparing the distribution before and after transformation helps us decide whether the transformation improved the feature for further Machine Learning processing.

```python
comparison = pd.DataFrame({
    "Original": df["Fare"],
    "Log": df["Fare_Log"],
    "Square Root": df["Fare_Sqrt"],
    "Power": df["Fare_Power"],
    "Yeo-Johnson": df["Fare_YeoJohnson"]
})

print("Skewness Comparison:")
print(comparison.skew())
```
